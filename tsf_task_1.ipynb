```python
# TASK 1: SIMPLE EXPONENTIAL SMOOTHING

# Step 1: Import libraries
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.datasets import get_rdataset
from statsmodels.tsa.holtwinters import SimpleExpSmoothing

# Step 2: Load AirPassengers dataset
data = get_rdataset("AirPassengers", "datasets").data

# Step 3: Prepare the time-series data
data['time'] = pd.date_range(start='1949-01-01',
                             periods=len(data),
                             freq='MS')

data.set_index('time', inplace=True)

# Rename the passenger column
data.rename(columns={'value': 'Passengers'}, inplace=True)

# Step 4: Create the time series
ts = data['Passengers']

# Step 5: Apply Simple Exponential Smoothing
model = SimpleExpSmoothing(ts)

# Step 6: Fit the model
fit = model.fit(
    smoothing_level=0.3,
    optimized=False
)

# Step 7: Get smoothed values
smoothed = fit.fittedvalues

# Step 8: Plot original and smoothed data
plt.figure(figsize=(12, 6))

plt.plot(ts,
         label='Original Data')

plt.plot(smoothed,
         label='Smoothed Data')

plt.title('Simple Exponential Smoothing of AirPassengers')
plt.xlabel('Time')
plt.ylabel('Number of Passengers')
plt.legend()
plt.grid()

plt.show()
```
